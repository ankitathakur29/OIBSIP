### Objective
The objective of this project is to perform Exploratory Data Analysis (EDA) on a retail sales dataset to identify sales trends, customer demographics, product category performance, and useful business insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("retail_sales_dataset.csv")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values in Each Column:")
df.isnull().sum()
```

```python
numerical_columns = df.select_dtypes(include="number")

print("Mean:")
print(numerical_columns.mean())

print("\nMedian:")
print(numerical_columns.median())

print("\nMode:")
print(numerical_columns.mode().iloc[0])

print("\nStandard Deviation:")
print(numerical_columns.std())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Start Date:", df["Date"].min())
print("End Date:", df["Date"].max())
```

```python
# Create a month column
df["Month"] = df["Date"].dt.to_period("M")

# Calculate monthly sales
monthly_sales = df.groupby("Month")["Total Amount"].sum()

monthly_sales
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

- Monthly sales fluctuate throughout the year.
- Sales show some months with higher and lower performance.
- January 2024 has very low sales because the dataset contains only the first day of January 2024, so it should not be compared directly with full months.
- The monthly sales trend can help the business identify high-performing and low-performing periods.

### Quarterly Sales Observation

- Sales performance varies across different quarters.
- Some quarters generate higher total sales than others.
- Quarterly analysis helps identify long-term sales patterns.
- These insights can help with sales planning and inventory management.

## Conclusion
